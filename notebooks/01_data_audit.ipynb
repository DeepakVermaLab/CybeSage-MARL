Audit dataset paths/manifests and CybORG import status.


```python
from pathlib import Path
import os
import sys

# If auto-detection fails, set this manually to the IMPLEMENTATION folder visible to this notebook kernel.
# Local VS Code kernel example:
# REPO_ROOT_OVERRIDE = "/Volumes/LocalDrive1/onlyscholar-projects/CybSage-MARL[DeepakSir]/IMPLEMENTATION"
# Colab example after copying the project to Drive:
# REPO_ROOT_OVERRIDE = "/content/drive/MyDrive/projects/implementation/IMPLEMENTATION"
REPO_ROOT_OVERRIDE = os.environ.get("CYBSAGE_REPO_ROOT", "")


def _candidate_roots():
    candidates = []
    if REPO_ROOT_OVERRIDE:
        candidates.append(Path(REPO_ROOT_OVERRIDE).expanduser())
    cwd = Path.cwd().resolve()
    candidates.extend([cwd, *cwd.parents])
    candidates.extend([
        Path("/Volumes/LocalDrive1/onlyscholar-projects/CybSage-MARL[DeepakSir]/IMPLEMENTATION"),
        Path.home() / "onlyscholar-projects/CybSage-MARL[DeepakSir]/IMPLEMENTATION",
        Path.home() / "CybSage-MARL[DeepakSir]/IMPLEMENTATION",
        Path("/content/drive/MyDrive/projects/implementation/IMPLEMENTATION"),
        Path("/content/drive/MyDrive/projects/implementation/CybSage-MARL[DeepakSir]/IMPLEMENTATION"),
        Path("/content/drive/MyDrive/IMPLEMENTATION"),
        Path("/content/drive/MyDrive/CybSage-MARL/IMPLEMENTATION"),
        Path("/content/drive/MyDrive/CybSAGE-MARL/IMPLEMENTATION"),
        Path("/content/drive/MyDrive/CybSage-MARL[DeepakSir]/IMPLEMENTATION"),
        Path("/content/IMPLEMENTATION"),
    ])
    return candidates


def _find_repo_root():
    checked = []
    for base in _candidate_roots():
        base = base.resolve() if base.exists() else base
        checked.append(str(base))
        if (base / "pyproject.toml").exists() and (base / "src" / "cybsage_marl").exists():
            return base
    msg = "\n".join(f"  - {p}" for p in checked)
    raise RuntimeError(
        "Could not find the CybSAGE IMPLEMENTATION repo root visible to this notebook kernel.\n\n"
        "Use a local VS Code kernel for local datasets, or mount/copy IMPLEMENTATION into Colab Drive.\n\n"
        "Checked paths:\n" + msg
    )

REPO_ROOT = _find_repo_root()
os.chdir(REPO_ROOT)
src_path = str(REPO_ROOT / "src")
if src_path not in sys.path:
    sys.path.insert(0, src_path)
print("Repository root:", REPO_ROOT)

!PYTHONPATH=src python scripts/run_stage.py audit
```
